# PARTE I: TALLER 5 (SERIES DE FOURIER)
## Objetivo:
Graficar la función periódica $f(x)$ superponiendo las sumas parciales armónicas para $N = 5$, $N = 10$ y $N = 20$ armónicos, identificando la cantidad mínima de armónicos ($n$) para lograr una aproximación fiel de la señal original.


### Sección 1 - Ejercicio 1: Pulso rectangular en $(-\pi, \pi)$
$$f(x) = \begin{cases} 0, & -\pi < x < 0 \\ 1, & 0 \le x < \pi \end{cases}, \quad f(x+2\pi) = f(x)$$

**Deducción Analítica:**
- $a_0 = \frac{1}{\pi}\int_0^\pi 1 dx = 1 \implies \frac{a_0}{2} = \frac{1}{2}$
- $a_n = \frac{1}{\pi}\int_0^\pi \cos(nx)dx = 0$
- $b_n = \frac{1}{\pi}\int_0^\pi \sin(nx)dx = \frac{1 - (-1)^n}{n\pi} = \begin{cases} \frac{2}{n\pi}, & n \text{ impar} \\ 0, & n \text{ par} \end{cases}$

$$f(x) \sim \frac{1}{2} + \frac{2}{\pi}\sum_{k=1}^{\infty}\frac{\sin((2k-1)x)}{2k-1}$$

**Teorema de Dirichlet en $x=0$:** $S(0) = \frac{f(0^+) + f(0^-)}{2} = \frac{1 + 0}{2} = 0.5$.


```python
x = np.linspace(-2*np.pi, 2*np.pi, 1000)

def f_orig_t5_1(x_val):
    p = (x_val + np.pi) % (2 * np.pi) - np.pi
    return np.where(p >= 0, 1.0, 0.0)

def fourier_t5_1(x_val, N):
    y = np.full_like(x_val, 0.5)
    for k in range(1, N + 1, 2):
        y += (2.0 / (k * np.pi)) * np.sin(k * x_val)
    return y

plt.figure()
plt.plot(x, f_orig_t5_1(x), 'k--', label='f(x) Original', linewidth=2)
for N in [5, 10, 20]:
    plt.plot(x, fourier_t5_1(x, N), label=f'N = {N} armónicos')
plt.scatter([0, np.pi, -np.pi], [0.5, 0.5, 0.5], color='red', zorder=5, label='Dirichlet (0.5)')
plt.title('Taller 5 - Ejercicio 1: Pulso Rectangular (N = 5, 10, 20)')
plt.xlabel('x (rad)')
plt.ylabel('f(x)')
plt.legend()
plt.show()
```

### Sección 1 - Ejercicio 4: Rampa lineal en $(-\pi, \pi)$
$$f(x) = x + \pi, \quad -\pi < x < \pi, \quad f(x+2\pi) = f(x)$$

**Deducción Analítica:**
- $a_0 = 2\pi \implies \frac{a_0}{2} = \pi$
- $a_n = 0 \quad \forall n \ge 1$
- $b_n = \frac{2(-1)^{n+1}}{n}$

$$f(x) \sim \pi + 2\sum_{n=1}^{\infty}\frac{(-1)^{n+1}}{n}\sin(nx)$$

**Teorema de Dirichlet en $x = \pm \pi$:** $S(\pm \pi) = \frac{2\pi + 0}{2} = \pi$.


```python
def f_orig_t5_4(x_val):
    p = (x_val + np.pi) % (2 * np.pi) - np.pi
    return p + np.pi

def fourier_t5_4(x_val, N):
    y = np.full_like(x_val, np.pi)
    for n in range(1, N + 1):
        y += (2.0 * ((-1)**(n+1)) / n) * np.sin(n * x_val)
    return y

plt.figure()
plt.plot(x, f_orig_t5_4(x), 'k--', label='f(x) Original: x + pi', linewidth=2)
for N in [5, 10, 20]:
    plt.plot(x, fourier_t5_4(x, N), label=f'N = {N} armónicos')
plt.scatter([np.pi, -np.pi], [np.pi, np.pi], color='red', zorder=5, label='Dirichlet en ±pi (y = pi)')
plt.title('Taller 5 - Ejercicio 4: Rampa Lineal (N = 5, 10, 20)')
plt.xlabel('x (rad)')
plt.ylabel('f(x)')
plt.legend()
plt.show()
```

### Sección 2 - Ejercicio 2: Serie de Cosenos $f(x) = x^2$ en $(-1, 1)$
Función Par $\implies b_n = 0$. $L = 1$.
- $a_0 = 2/3 \implies \frac{a_0}{2} = \frac{1}{3}$
- $a_n = \frac{4(-1)^n}{n^2\pi^2}$

$$f(x) = \frac{1}{3} + \frac{4}{\pi^2}\sum_{n=1}^{\infty}\frac{(-1)^n}{n^2}\cos(n\pi x)$$

**Convergencia:** Tasa $\mathcal{O}(1/n^2)$ acelerada; continua en todo $\mathbb{R}$ sin fenómeno de Gibbs.


```python
x_sc = np.linspace(-2.5, 2.5, 1000)

def f_orig_t5_sc2(x_val):
    p = (x_val + 1) % 2 - 1
    return p**2

def fourier_t5_sc2(x_val, N):
    y = np.full_like(x_val, 1.0 / 3.0)
    for n in range(1, N + 1):
        y += (4.0 * ((-1)**n) / ((n * np.pi)**2)) * np.cos(n * np.pi * x_val)
    return y

plt.figure()
plt.plot(x_sc, f_orig_t5_sc2(x_sc), 'k--', label='f(x) Original: x^2 (Par)', linewidth=2)
for N in [5, 10, 20]:
    plt.plot(x_sc, fourier_t5_sc2(x_sc, N), label=f'N = {N} armónicos')
plt.title('Taller 5 - Ejercicio 2 (Cosenos): x^2 en (-1, 1)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.show()
```

### Sección 2 - Ejercicio 4: Serie de Senos $f(x) = x^3$ en $(-\pi, \pi)$
Función Impar $\implies a_0 = a_n = 0$. $L = \pi$.
- $b_n = 2(-1)^{n+1}\left(\frac{\pi^2}{n} - \frac{6}{n^3}\right)$

$$f(x) = 2\sum_{n=1}^{\infty}(-1)^{n+1}\left(\frac{\pi^2}{n} - \frac{6}{n^3}\right)\sin(nx)$$


```python
def f_orig_t5_sc4(x_val):
    p = (x_val + np.pi) % (2 * np.pi) - np.pi
    return p**3

def fourier_t5_sc4(x_val, N):
    y = np.zeros_like(x_val)
    for n in range(1, N + 1):
        bn = 2.0 * ((-1)**(n+1)) * ((np.pi**2)/n - 6.0/(n**3))
        y += bn * np.sin(n * x_val)
    return y

plt.figure()
plt.plot(x, f_orig_t5_sc4(x), 'k--', label='f(x) Original: x^3 (Impar)', linewidth=2)
for N in [5, 10, 20]:
    plt.plot(x, fourier_t5_sc4(x, N), label=f'N = {N} armónicos')
plt.scatter([np.pi, -np.pi], [0, 0], color='red', zorder=5, label='Dirichlet en ±pi (y = 0)')
plt.title('Taller 5 - Ejercicio 4 (Senos): x^3 en (-pi, pi)')
plt.xlabel('x (rad)')
plt.ylabel('f(x)')
plt.legend()
plt.show()
```

### Sección 3 - Ejercicio 1: Serie Compleja en Onda Cuadrada en $(-2, 2)$
$T = 4, \; \omega_0 = \pi/2$.
- $c_0 = 0$
- $c_n = \begin{cases} -\frac{2i}{n\pi}, & n \text{ impar} \\ 0, & n \text{ par} \end{cases}$

$$f(x) = \sum_{n=-\infty, n\neq 0}^{\infty}\left(-\frac{2i}{n\pi}\right)e^{in\pi x/2} = \frac{4}{\pi}\sum_{k=1}^{\infty}\frac{\sin((2k-1)\pi x / 2)}{2k-1}$$


```python
x_comp = np.linspace(-4, 4, 1000)

def f_orig_t5_c1(x_val):
    p = (x_val + 2) % 4 - 2
    return np.where(p > 0, 1.0, -1.0)

def fourier_t5_c1(x_val, N):
    y = np.zeros(len(x_val), dtype=complex)
    for n in range(-N, N + 1):
        if n % 2 != 0:
            cn = -2.0j / (n * np.pi)
            y += cn * np.exp(1j * n * np.pi * x_val / 2.0)
    return y.real

plt.figure()
plt.plot(x_comp, f_orig_t5_c1(x_comp), 'k--', label='Onda Cuadrada Original', linewidth=2)
for N in [5, 10, 20]:
    plt.plot(x_comp, fourier_t5_c1(x_comp, N), label=f'N = {N} armónicos complejos')
plt.scatter([0, 2, -2], [0, 0, 0], color='red', zorder=5, label='Dirichlet (0)')
plt.title('Taller 5 - Ejercicio 1 (Complejas): Onda Cuadrada en (-2, 2)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.show()
```

### Sección 3 - Ejercicio 6: Serie Compleja Par Exponencial $f(x) = e^{-|x|}$ en $(-1, 1)$
$T = 2, \; \omega_0 = \pi$.
- $c_0 = 1 - e^{-1}$
- $c_n = \frac{1 - e^{-1}(-1)^n}{1 + n^2\pi^2}$

$$f(x) = (1 - e^{-1}) + 2\sum_{n=1}^{\infty}\frac{1 - e^{-1}(-1)^n}{1 + n^2\pi^2}\cos(n\pi x)$$


```python
def f_orig_t5_c6(x_val):
    p = (x_val + 1) % 2 - 1
    return np.exp(-np.abs(p))

def fourier_t5_c6(x_val, N):
    e_inv = np.exp(-1)
    y = np.full_like(x_val, 1.0 - e_inv)
    for n in range(1, N + 1):
        cn = (1.0 - e_inv * ((-1)**n)) / (1.0 + (n * np.pi)**2)
        y += 2.0 * cn * np.cos(n * np.pi * x_val)
    return y

plt.figure()
plt.plot(x_sc, f_orig_t5_c6(x_sc), 'k--', label='f(x) Original: exp(-|x|)', linewidth=2)
for N in [5, 10, 20]:
    plt.plot(x_sc, fourier_t5_c6(x_sc, N), label=f'N = {N} armónicos')
plt.title('Taller 5 - Ejercicio 6 (Complejas): exp(-|x|) en (-1, 1)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.show()
```

# PARTE II: TALLER 6 (INTEGRALES DE FOURIER)
## Objetivo:
Evaluar numéricamente la representación integral truncada con frecuencia de corte finita $b$ ($b = 5, 20, 50$):
$$f_b(x) = \frac{1}{\pi} \int_0^b [A(\alpha)\cos(\alpha x) + B(\alpha)\sin(\alpha x)] d\alpha$$


### Sección 1 - Ejercicio 2: Integral General de Fourier
$$f(x) = \begin{cases} 4, & \pi < x < 2\pi \\ 0, & \text{en otro caso} \end{cases}$$

- $A(\alpha) = \frac{4[\sin(2\pi\alpha) - \sin(\pi\alpha)]}{\alpha}$
- $B(\alpha) = \frac{4[\cos(\pi\alpha) - \cos(2\pi\alpha)]}{\alpha}$

**Dirichlet en $x = \pi$ y $x = 2\pi$:** $\frac{4 + 0}{2} = 2$.


```python
def f_orig_t6_2(x_val):
    return np.where((x_val > np.pi) & (x_val < 2*np.pi), 4.0, 0.0)

def f_b_t6_2(x_val, b):
    def integrand(alpha):
        if alpha == 0:
            return 4.0 * np.pi
        A = 4.0 * (np.sin(2*np.pi*alpha) - np.sin(np.pi*alpha)) / alpha
        B = 4.0 * (np.cos(np.pi*alpha) - np.cos(2*np.pi*alpha)) / alpha
        return A * np.cos(alpha * x_val) + B * np.sin(alpha * x_val)
    res, _ = quad(integrand, 0, b, limit=150)
    return res / np.pi

x_t6_2 = np.linspace(-2, 3*np.pi, 250)
plt.figure()
plt.plot(x_t6_2, f_orig_t6_2(x_t6_2), 'k--', label='f(x) Original (Pulso 4 en [pi, 2pi])', linewidth=2)
for b_cut in [5, 20, 50]:
    y_approx = [f_b_t6_2(xi, b_cut) for xi in x_t6_2]
    plt.plot(x_t6_2, y_approx, label=f'Corte b = {b_cut} rad/u')
plt.scatter([np.pi, 2*np.pi], [2, 2], color='red', zorder=5, label='Dirichlet (y = 2)')
plt.title('Taller 6 - Ejercicio 2: Integral General de Fourier')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.show()
```

### Sección 1 - Ejercicio 5: Integral Par $f(x) = e^{-|x|}$ en $(-\infty, \infty)$
Función Par $\implies B(\alpha) = 0$.
- $A(\alpha) = \frac{2}{1 + \alpha^2}$

$$f_b(x) = \frac{2}{\pi}\int_0^b \frac{\cos(\alpha x)}{1 + \alpha^2}d\alpha$$

**Cúspide en $x=0$:** $f_b(0) = \frac{2}{\pi}\arctan(b) \to 1$ cuando $b \to \infty$.


```python
def f_orig_t6_5(x_val):
    return np.exp(-np.abs(x_val))

def f_b_t6_5(x_val, b):
    def integrand(alpha):
        return np.cos(alpha * x_val) / (1.0 + alpha**2)
    res, _ = quad(integrand, 0, b)
    return (2.0 / np.pi) * res

x_t6_5 = np.linspace(-5, 5, 250)
plt.figure()
plt.plot(x_t6_5, f_orig_t6_5(x_t6_5), 'k--', label='f(x) Original: exp(-|x|)', linewidth=2)
for b_cut in [5, 20, 50]:
    y_approx = [f_b_t6_5(xi, b_cut) for xi in x_t6_5]
    plt.plot(x_t6_5, y_approx, label=f'Corte b = {b_cut} rad/u')
plt.title('Taller 6 - Ejercicio 5: Integral Par exp(-|x|)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.show()
```

### Sección 2 - Ejercicio 1: Integral de Senos (Pulso Antisimétrico)
$$f(x) = \begin{cases} -5, & -1 < x < 0 \\ 5, & 0 < x < 1 \\ 0, & |x| > 1 \end{cases}$$

Función Impar $\implies A(\alpha) = 0$.
- $B(\alpha) = \frac{10(1 - \cos\alpha)}{\alpha}$

$$f_b(x) = \frac{1}{\pi}\int_0^b \frac{10(1 - \cos\alpha)}{\alpha}\sin(\alpha x)d\alpha$$

**Dirichlet:** Salto en $x=0 \implies 0$, en $x = 1 \implies 2.5$, en $x = -1 \implies -2.5$.


```python
def f_orig_t6_s1(x_val):
    return np.where((x_val > 0) & (x_val < 1), 5.0, np.where((x_val > -1) & (x_val < 0), -5.0, 0.0))

def f_b_t6_s1(x_val, b):
    def integrand(alpha):
        if alpha == 0:
            return 0.0
        return (10.0 * (1.0 - np.cos(alpha)) / alpha) * np.sin(alpha * x_val)
    res, _ = quad(integrand, 0, b, limit=150)
    return res / np.pi

x_t6_s1 = np.linspace(-3, 3, 250)
plt.figure()
plt.plot(x_t6_s1, f_orig_t6_s1(x_t6_s1), 'k--', label='f(x) Original (Impar: ±5)', linewidth=2)
for b_cut in [5, 20, 50]:
    y_approx = [f_b_t6_s1(xi, b_cut) for xi in x_t6_s1]
    plt.plot(x_t6_s1, y_approx, label=f'Corte b = {b_cut} rad/u')
plt.scatter([-1, 0, 1], [-2.5, 0, 2.5], color='red', zorder=5, label='Dirichlet en saltos')
plt.title('Taller 6 - Ejercicio 1 (Senos): Pulso Antisimétrico')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.show()
```

### Sección 2 - Ejercicio 2: Integral de Cosenos (Pulso Par Simétrico)
$$f(x) = \begin{cases} \pi, & 1 < |x| < 2 \\ 0, & \text{en otro caso} \end{cases}$$

Función Par $\implies B(\alpha) = 0$.
- $A(\alpha) = \frac{2\pi(\sin(2\alpha) - \sin\alpha)}{\alpha}$

$$f_b(x) = 2\int_0^b \frac{\sin(2\alpha) - \sin\alpha}{\alpha}\cos(\alpha x)d\alpha$$

**Dirichlet:** Saltos en $x = \pm 1$ y $x = \pm 2 \implies \frac{\pi + 0}{2} = \frac{\pi}{2} \approx 1.5708$.


```python
def f_orig_t6_c2(x_val):
    ax = np.abs(x_val)
    return np.where((ax > 1.0) & (ax < 2.0), np.pi, 0.0)

def f_b_t6_c2(x_val, b):
    def integrand(alpha):
        if alpha == 0:
            return 1.0 * np.cos(0)
        return ((np.sin(2.0*alpha) - np.sin(alpha)) / alpha) * np.cos(alpha * x_val)
    res, _ = quad(integrand, 0, b, limit=150)
    return 2.0 * res

x_t6_c2 = np.linspace(-4, 4, 250)
plt.figure()
plt.plot(x_t6_c2, f_orig_t6_c2(x_t6_c2), 'k--', label='f(x) Original (Ventanas par = pi)', linewidth=2)
for b_cut in [5, 20, 50]:
    y_approx = [f_b_t6_c2(xi, b_cut) for xi in x_t6_c2]
    plt.plot(x_t6_c2, y_approx, label=f'Corte b = {b_cut} rad/u')
plt.scatter([-2, -1, 1, 2], [np.pi/2]*4, color='red', zorder=5, label='Dirichlet (pi/2)')
plt.title('Taller 6 - Ejercicio 2 (Cosenos): Pulso Par Simétrico')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.show()
```

# CONCLUSIONES TEÓRICO-PRÁCTICAS

1. **Velocidad de convergencia:** Se evidencia que las funciones continuas con derivadas seccionales ($x^2$, con tasa $\mathcal{O}(1/n^2)$) requieren un número significativamente menor de armónicos ($N \approx 5$) para alcanzar un error relativo menor al 1%, a diferencia de las funciones discontinuas (pulso rectangular, rampa lineal y onda cuadrada, con tasa $\mathcal{O}(1/n)$), que demandan $N \ge 20$ armónicos.

2. **Fenómeno de Gibbs:** En todas las discontinuidades de salto finito se manifestó un sobreimpulso asintótico constante del $\approx 8.95\%$. Al aumentar $N$ o $b$, la oscilación no disminuye en amplitud pico; únicamente se comprime espacialmente hacia el salto.

3. **Verificación del Teorema de Dirichlet:** Tanto en las series periódicas como en las integrales no periódicas truncadas, la aproximación evaluada en cualquier punto de discontinuidad $x_0$ converge con precisión al promedio aritmético exacto de los límites laterales: $\frac{f(x_0^+) + f(x_0^-)}{2}$.

4. **Atenuación del rizado espectral en Integrales de Fourier:** Al incrementar la frecuencia de corte $b$ de $5 \to 20 \to 50$ rad/u, las oscilaciones en los tramos suaves se atenúan proporcionalmente a $1/b$, reconstruyendo la señal no periódica en todo el dominio continuo.
