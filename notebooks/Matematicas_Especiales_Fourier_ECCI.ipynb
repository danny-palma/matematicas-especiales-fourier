## 3. Motor de Series de Fourier Trigonométricas (Taller 5, Secciones 1 y 2)

Calcula la suma parcial $S_n(x)$ en el intervalo base de semiperiodo $p = \frac{b - a}{2}$:

$$S_n(x) = \frac{a_0}{2} + \sum_{k=1}^{n} \left[ a_k \cos\left(\frac{k\pi x}{p}\right) + b_k \sin\left(\frac{k\pi x}{p}\right) \right]$$

- **`modo_simetria="general"`:** Calcula $a_0$, $a_k$ y $b_k$.
- **`modo_simetria="cosenos"` (función par):** Calcula solo $a_0$ y $a_k$ ($b_k = 0$).
- **`modo_simetria="senos"` (función impar):** Calcula solo $b_k$ ($a_0 = a_k = 0$).

```python
def aproximar_serie_fourier_trigonometrica(arreglo_x, tramos_funcion, cantidad_terminos_n, modo_simetria="general"):
    inicio_periodo = tramos_funcion[0][0]
    fin_periodo = tramos_funcion[-1][1]
    semiperiodo_p = (fin_periodo - inicio_periodo) / 2.0

    coeficiente_a0 = 0.0
    if modo_simetria in ("general", "cosenos"):
        for limite_inferior, limite_superior, expresion_texto in tramos_funcion:
            integral_tramo, _ = quad(lambda x: evaluar_expresion_matematica(expresion_texto, x), limite_inferior, limite_superior)
            coeficiente_a0 += (1.0 / semiperiodo_p) * integral_tramo

    suma_parcial_sn = np.full_like(arreglo_x, coeficiente_a0 / 2.0, dtype=float)

    for armonico_n in range(1, cantidad_terminos_n + 1):
        frecuencia_n = (armonico_n * np.pi) / semiperiodo_p
        coeficiente_an = 0.0
        coeficiente_bn = 0.0

        for limite_inferior, limite_superior, expresion_texto in tramos_funcion:
            if modo_simetria in ("general", "cosenos"):
                integral_cos, _ = quad(
                    lambda x: evaluar_expresion_matematica(expresion_texto, x) * np.cos(frecuencia_n * x),
                    limite_inferior, limite_superior
                )
                coeficiente_an += (1.0 / semiperiodo_p) * integral_cos

            if modo_simetria in ("general", "senos"):
                integral_sen, _ = quad(
                    lambda x: evaluar_expresion_matematica(expresion_texto, x) * np.sin(frecuencia_n * x),
                    limite_inferior, limite_superior
                )
                coeficiente_bn += (1.0 / semiperiodo_p) * integral_sen

        suma_parcial_sn += coeficiente_an * np.cos(frecuencia_n * arreglo_x) + coeficiente_bn * np.sin(frecuencia_n * arreglo_x)

    return suma_parcial_sn
```

## 4. Motor de Series Complejas de Fourier (Taller 5, Sección 3)

Utiliza la forma exponencial compleja sumando desde $-n$ hasta $+n$:

$$c_k = \frac{1}{2p} \int_{-p}^{p} f(x) e^{-i k \pi x / p} dx, \qquad S_n(x) = \sum_{k=-n}^{n} c_k e^{i k \pi x / p}$$

Como `quad` integra funciones reales, la integral de $f(x)e^{-ik\pi x/p}$ se separa en su parte real ($\cos$) y su parte imaginaria ($\sin$), y al final se toma `np.real(suma_compleja)`.

```python
def aproximar_serie_compleja_fourier(arreglo_x, tramos_funcion, cantidad_terminos_n):
    inicio_periodo = tramos_funcion[0][0]
    fin_periodo = tramos_funcion[-1][1]
    semiperiodo_p = (fin_periodo - inicio_periodo) / 2.0
    suma_compleja = np.zeros_like(arreglo_x, dtype=complex)

    for indice_k in range(-cantidad_terminos_n, cantidad_terminos_n + 1):
        frecuencia_k = (indice_k * np.pi) / semiperiodo_p
        parte_real_ck = 0.0
        parte_imaginaria_ck = 0.0

        for limite_inferior, limite_superior, expresion_texto in tramos_funcion:
            integral_real, _ = quad(
                lambda x: evaluar_expresion_matematica(expresion_texto, x) * np.cos(-frecuencia_k * x),
                limite_inferior, limite_superior
            )
            integral_imag, _ = quad(
                lambda x: evaluar_expresion_matematica(expresion_texto, x) * np.sin(-frecuencia_k * x),
                limite_inferior, limite_superior
            )
            parte_real_ck += integral_real
            parte_imaginaria_ck += integral_imag

        coeficiente_ck = (parte_real_ck + 1j * parte_imaginaria_ck) / (2.0 * semiperiodo_p)
        suma_compleja += coeficiente_ck * np.exp(1j * frecuencia_k * arreglo_x)

    return np.real(suma_compleja)
```

## 5. Motor de Integral de Fourier Truncada (Taller 6, Secciones 1 y 2)

Para funciones no periódicas en $(-\infty, \infty)$, se evalúa numéricamente la integral truncada hasta una frecuencia máxima $\alpha = b$:

$$F_b(x) = \frac{1}{\pi} \int_{0}^{b} \left[ A(\alpha)\cos(\alpha x) + B(\alpha)\sin(\alpha x) \right] d\alpha$$

También incluye el selector `calcular_aproximacion_segun_estilo` que enruta cada ejercicio a uno de los **5 estilos** del curso.

```python
def aproximar_integral_truncada_fourier(arreglo_x, tramos_funcion, frecuencia_maxima_b, modo_simetria="general"):
    malla_alfa = np.linspace(1e-5, float(frecuencia_maxima_b), 250)
    coeficientes_A_alfa = np.zeros_like(malla_alfa)
    coeficientes_B_alfa = np.zeros_like(malla_alfa)

    for indice, valor_alfa in enumerate(malla_alfa):
        for limite_inferior, limite_superior, expresion_texto in tramos_funcion:
            if modo_simetria in ("general", "cosenos"):
                int_cos, _ = quad(
                    lambda t: evaluar_expresion_matematica(expresion_texto, t) * np.cos(valor_alfa * t),
                    limite_inferior, limite_superior
                )
                coeficientes_A_alfa[indice] += int_cos

            if modo_simetria in ("general", "senos"):
                int_sen, _ = quad(
                    lambda t: evaluar_expresion_matematica(expresion_texto, t) * np.sin(valor_alfa * t),
                    limite_inferior, limite_superior
                )
                coeficientes_B_alfa[indice] += int_sen

    matriz_alfa_x = np.outer(malla_alfa, arreglo_x)
    integrando_frecuencia = (
        coeficientes_A_alfa[:, None] * np.cos(matriz_alfa_x)
        + coeficientes_B_alfa[:, None] * np.sin(matriz_alfa_x)
    )
    integrar_trapecio = getattr(np, "trapezoid", getattr(np, "trapz", None))
    return (1.0 / np.pi) * integrar_trapecio(integrando_frecuencia, malla_alfa, axis=0)


def calcular_aproximacion_segun_estilo(arreglo_x, configuracion_ejercicio, parametro_n_o_b):
    estilo = configuracion_ejercicio["estilo"]
    tramos = configuracion_ejercicio["tramos"]

    if estilo == "serie_general":
        return aproximar_serie_fourier_trigonometrica(arreglo_x, tramos, parametro_n_o_b, "general")
    if estilo == "serie_cosenos":
        return aproximar_serie_fourier_trigonometrica(arreglo_x, tramos, parametro_n_o_b, "cosenos")
    if estilo == "serie_senos":
        return aproximar_serie_fourier_trigonometrica(arreglo_x, tramos, parametro_n_o_b, "senos")
    if estilo == "serie_compleja":
        return aproximar_serie_compleja_fourier(arreglo_x, tramos, parametro_n_o_b)
    if estilo == "integral_general":
        return aproximar_integral_truncada_fourier(arreglo_x, tramos, parametro_n_o_b, "general")
    if estilo == "integral_cosenos":
        return aproximar_integral_truncada_fourier(arreglo_x, tramos, parametro_n_o_b, "cosenos")
    return aproximar_integral_truncada_fourier(arreglo_x, tramos, parametro_n_o_b, "senos")
```

## 6. Graficación Comparativa y Puntos de Discontinuidad

La función `graficar_configuracion` construye el lienzo comparativo sobre **1000 puntos**:
- **Línea azul punteada (`--`):** Función original $f(x)$.
- **Línea roja continua (`-`):** Aproximación de Fourier para el valor actual de `n` o `b` (o tres curvas simultáneas para `5`, `10` y `20` cuando se activa la casilla de la *Additional Activity*).
- **Franja gris:** Señala el intervalo base de definición frente al dominio extendido.
- **Puntos negros:** Marcan el promedio de los límites laterales $\frac{f(x^+) + f(x^-)}{2}$ en las discontinuidades de salto.

```python
def graficar_configuracion(nombre_ejercicio, configuracion, terminos_n_o_b=5, comparar_5_10_20=False):
    es_serie_periodica = configuracion["estilo"].startswith("serie")
    x_izquierdo, x_derecho = configuracion["rango_grafica"]
    arreglo_lineal_x = np.linspace(x_izquierdo, x_derecho, 1000)

    curva_original = evaluar_funcion_original(arreglo_lineal_x, configuracion["tramos"], es_periodica=es_serie_periodica)

    plt.figure(figsize=(11.5, 5.2))
    plt.plot(arreglo_lineal_x, curva_original, color="royalblue", linestyle="--", linewidth=2.0, label="Función Original f(x)")

    if comparar_5_10_20:
        for valor_comparativo, color_curva in [(5, "orange"), (10, "seagreen"), (20, "red")]:
            curva_aprox = calcular_aproximacion_segun_estilo(arreglo_lineal_x, configuracion, valor_comparativo)
            plt.plot(arreglo_lineal_x, curva_aprox, color=color_curva, linewidth=1.8, label=f"Aproximación Fourier con parámetro = {valor_comparativo}")
    else:
        curva_roja_principal = calcular_aproximacion_segun_estilo(arreglo_lineal_x, configuracion, terminos_n_o_b)
        plt.plot(arreglo_lineal_x, curva_roja_principal, color="red", linewidth=2.3, label=f"Aproximación Fourier (n / b = {terminos_n_o_b})")

    if configuracion["discontinuidades"]:
        x_saltos = [punto[0] for punto in configuracion["discontinuidades"]]
        y_medios = [punto[1] for punto in configuracion["discontinuidades"]]
        plt.scatter(x_saltos, y_medios, color="black", zorder=5, s=45, label="Valor medio en discontinuidad [f(x+)+f(x-)]/2")

    inicio_base = configuracion["tramos"][0][0]
    fin_base = configuracion["tramos"][-1][1]
    plt.axvspan(inicio_base, fin_base, color="gray", alpha=0.12, label=f"Intervalo Base [{inicio_base:.2f}, {fin_base:.2f}]")

    plt.title(f"{nombre_ejercicio} | Estilo: {configuracion['estilo']}", fontsize=12, fontweight="bold")
    plt.xlabel("Dominio x (1000 puntos equiespaciados)", fontsize=11)
    plt.ylabel("f(x) vs Aproximación de Fourier", fontsize=11)
    plt.axhline(0, color="black", linewidth=0.8)
    plt.axvline(0, color="black", linewidth=0.8)
    plt.grid(True, linestyle=":", alpha=0.7)
    plt.legend(loc="best", fontsize=9)
    plt.tight_layout()
    plt.show()
```

```python
opciones_menu = list(CATALOGO_EJERCICIOS_TALLERES.keys()) + ["0. [MODO DINÁMICO] Ingresar otra función por tramos"]

selector_ejercicio = widgets.Dropdown(options=opciones_menu, value=opciones_menu[4], description="Ejercicio:", layout=widgets.Layout(width="680px"))
deslizador_terminos = widgets.IntSlider(min=1, max=40, step=1, value=5, description="Iteración n/b:")
casilla_comparativa = widgets.Checkbox(value=False, description="Comparar 5, 10 y 20 (Additional Activity)")

selector_estilo_dinamico = widgets.Dropdown(
    options=["serie_general", "serie_cosenos", "serie_senos", "serie_compleja", "integral_general", "integral_cosenos", "integral_senos"],
    value="serie_general",
    description="Estilo:"
)
campo_cantidad_partes = widgets.BoundedIntText(value=2, min=1, max=10, step=1, description="Partes:")
campo_x_izquierdo = widgets.Text(value="-6", description="x mínimo:")
campo_x_derecho = widgets.Text(value="6", description="x máximo:")
boton_aplicar_dinamico = widgets.Button(description="Graficar Función Dinámica", button_style="primary")
contenedor_filas_tramos = widgets.VBox([])
panel_modo_dinamico = widgets.VBox([], layout=widgets.Layout(display="none", border="1px solid #ccc", padding="10px", margin="6px 0"))
salida_grafica = widgets.Output()

valores_tramos_iniciales = [("-2", "0", "-1"), ("0", "2", "1")]
filas_tramos_activas = []


def reconstruir_filas_tramos(_=None):
    global filas_tramos_activas
    filas_tramos_activas = []
    widgets_filas = []
    for indice in range(campo_cantidad_partes.value):
        lim_inf_def, lim_sup_def, expr_def = valores_tramos_iniciales[indice] if indice < len(valores_tramos_iniciales) else ("0", "1", "x")
        caja_inf = widgets.Text(value=lim_inf_def, description=f"Tramo {indice + 1} Inf:", layout=widgets.Layout(width="180px"))
        caja_sup = widgets.Text(value=lim_sup_def, description="Sup:", layout=widgets.Layout(width="150px"))
        caja_expr = widgets.Text(value=expr_def, description="f(x):", layout=widgets.Layout(width="260px"))
        filas_tramos_activas.append((caja_inf, caja_sup, caja_expr))
        widgets_filas.append(widgets.HBox([caja_inf, caja_sup, caja_expr]))
    contenedor_filas_tramos.children = widgets_filas


def obtener_configuracion_dinamica():
    tramos_usuario = [
        (evaluar_limite_numerico(c_inf.value), evaluar_limite_numerico(c_sup.value), c_expr.value.strip())
        for c_inf, c_sup, c_expr in filas_tramos_activas
    ]
    return {
        "estilo": selector_estilo_dinamico.value,
        "tramos": tramos_usuario,
        "rango_grafica": (evaluar_limite_numerico(campo_x_izquierdo.value), evaluar_limite_numerico(campo_x_derecho.value)),
        "discontinuidades": []
    }


def actualizar_vista(_=None):
    es_dinamico = selector_ejercicio.value.startswith("0.")
    panel_modo_dinamico.layout.display = "flex" if es_dinamico else "none"
    with salida_grafica:
        clear_output(wait=True)
        configuracion = obtener_configuracion_dinamica() if es_dinamico else CATALOGO_EJERCICIOS_TALLERES[selector_ejercicio.value]
        graficar_configuracion(selector_ejercicio.value, configuracion, deslizador_terminos.value, casilla_comparativa.value)


reconstruir_filas_tramos()
panel_modo_dinamico.children = [
    widgets.HBox([selector_estilo_dinamico, campo_cantidad_partes]),
    contenedor_filas_tramos,
    widgets.HBox([campo_x_izquierdo, campo_x_derecho, boton_aplicar_dinamico])
]

campo_cantidad_partes.observe(reconstruir_filas_tramos, names="value")
selector_ejercicio.observe(actualizar_vista, names="value")
deslizador_terminos.observe(actualizar_vista, names="value")
casilla_comparativa.observe(actualizar_vista, names="value")
boton_aplicar_dinamico.on_click(actualizar_vista)

display(widgets.VBox([selector_ejercicio, deslizador_terminos, casilla_comparativa, panel_modo_dinamico, salida_grafica]))
actualizar_vista()
```
